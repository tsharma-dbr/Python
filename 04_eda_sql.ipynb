# 04 — Exploratory Data Analysis with SQL
SQL is used as an auditable analytical layer for filtering, aggregation, grouping, ordering, limiting,
and subqueries.

```python
import pandas as pd
import sqlite3

DATA_URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv"
df = pd.read_csv(DATA_URL)

conn = sqlite3.connect(":memory:")
df.to_sql("SPACEXTBL", conn, index=False, if_exists="replace")
```

```python
q1 = 'SELECT * FROM SPACEXTBL LIMIT 20'
display(pd.read_sql(q1, conn))
```

```python
q2 = 'SELECT MIN(PayloadMass) AS Minimum_Payload_Mass FROM SPACEXTBL'
display(pd.read_sql(q2, conn))
```

```python
q3 = '''
SELECT SUM(PayloadMass) AS Total_Payload_Mass
FROM SPACEXTBL
'''
display(pd.read_sql(q3, conn))
```

```python
q4 = '''
SELECT LaunchSite, COUNT(*) AS Mission_Outcome_Count
FROM SPACEXTBL
GROUP BY LaunchSite
ORDER BY Mission_Outcome_Count DESC
'''
display(pd.read_sql(q4, conn))
```

```python
# Additional SQL examples used in the capstone.
queries = {
    "launch_site_count": '''
        SELECT LaunchSite, COUNT(*) AS launch_count
        FROM SPACEXTBL
        GROUP BY LaunchSite
        ORDER BY launch_count DESC
    ''',
    "payload_by_booster": '''
        SELECT BoosterVersion, SUM(PayloadMass) AS Total_Payload_Mass
        FROM SPACEXTBL
        GROUP BY BoosterVersion
        ORDER BY Total_Payload_Mass DESC
    '''
}

for name, query in queries.items():
    print(name)
    display(pd.read_sql(query, conn))
```

**Note:** exact column names can vary slightly between course dataset versions. Before running a query,
inspect `df.columns` and adjust names only when the local CSV uses a different spelling.
