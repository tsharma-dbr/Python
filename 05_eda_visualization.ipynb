# 05 — Exploratory Data Analysis with Visualization
This notebook uses Pandas, Matplotlib and Seaborn to explore launch-site, payload, orbit and landing-outcome patterns.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv"
df = pd.read_csv(DATA_URL)
df.head()
```

```python
plt.figure(figsize=(10,5))
sns.catplot(
    y="LaunchSite",
    x="FlightNumber",
    hue="Class",
    data=df,
    aspect=1
)
plt.ylabel("Launch Site", fontsize=15)
plt.xlabel("Flight Number", fontsize=15)
plt.show()
```

The capstone uses a Seaborn categorical scatter/strip view for `FlightNumber` versus `LaunchSite`,
with `Class` as the hue. A standard bar chart is particularly appropriate for categorical summaries.

```python
plt.figure(figsize=(9,5))
success_by_site = df.groupby("LaunchSite")["Class"].mean().sort_values(ascending=False)
success_by_site.plot(kind="bar")
plt.ylabel("Landing success rate")
plt.xlabel("Launch Site")
plt.title("Landing Success Rate by Launch Site")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.scatterplot(
    data=df,
    x="FlightNumber",
    y="PayloadMass",
    hue="Class"
)
plt.title("Flight Number vs Payload Mass by Landing Class")
plt.tight_layout()
plt.show()
```
