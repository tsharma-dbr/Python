# 03 — SpaceX Data Wrangling
This notebook prepares the launch-level dataset for exploratory analysis and classification.

```python
import pandas as pd
import numpy as np

DATA_URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv"
df = pd.read_csv(DATA_URL)
df.head()
```

```python
# Inspect structure and missing values.
print(df.info())
missing = df.isna().sum().sort_values(ascending=False)
display(missing.to_frame("MissingValues").head(15))
```

```python
# Keep the course modeling fields.
features = df[[
    "FlightNumber", "PayloadMass", "Orbit", "LaunchSite", "Flights",
    "GridFins", "Reused", "Legs", "LandingPad", "Block",
    "ReusedCount", "Serial"
]].copy()

# Binary landing target.
Y = df["Class"].astype(int)

# One-hot encoding of categorical variables.
features_one_hot = pd.get_dummies(
    features,
    columns=["Orbit", "LaunchSite", "LandingPad", "Serial"],
    dummy_na=False
)

print("Encoded feature columns:", features_one_hot.shape[1])
features_one_hot.head()
```

**Verified course result:** after one-hot encoding `Orbit`, `LaunchSite`, `LandingPad`, and `Serial`,
the feature matrix contains **80 columns**.
