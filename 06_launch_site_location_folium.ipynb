# 06 — Launch-Site Location Analysis with Folium
The map stage visualizes launch locations, supports point markers and uses MarkerCluster for dense locations.

```python
import pandas as pd
import folium
from folium.plugins import MarkerCluster

DATA_URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_geo.csv"
geo = pd.read_csv(DATA_URL)
geo.head()
```

```python
center = [geo["Latitude"].mean(), geo["Longitude"].mean()]
m = folium.Map(location=center, zoom_start=4)

cluster = MarkerCluster().add_to(m)

for _, row in geo.iterrows():
    folium.Marker(
        location=[row["Latitude"], row["Longitude"]],
        popup=str(row.get("Launch Site", "SpaceX Launch Site"))
    ).add_to(cluster)

m
```

### Folium API patterns used in the capstone

- `object.add_to(map)`
- `map.add_child(object)`
- `MarkerCluster` for multiple nearby markers
