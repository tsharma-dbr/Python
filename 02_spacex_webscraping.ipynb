# 02 — SpaceX Data Collection with Web Scraping
The web-scraping stage retrieves Falcon 9 launch-history information from the Wikipedia launch-list page
and demonstrates BeautifulSoup parsing.

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

wiki_url = "https://en.wikipedia.org/wiki/List_of_Falcon_9_and_Falcon_Heavy_launches"
html = requests.get(
    wiki_url,
    headers={"User-Agent": "Mozilla/5.0 (compatible; Capstone/1.0)"},
    timeout=30
).text

soup = BeautifulSoup(html, "html.parser")
print(soup.title)
```

**Verified course output for `soup.title`:**

`<title> List of Falcon 9 and Falcon Heavy launches - Wikipedia </title>`

```python
# Inspect the main launch-history tables.
tables = pd.read_html(html)
print("Tables found:", len(tables))
for i, t in enumerate(tables[:8]):
    print(i, t.shape)
```

### Parsing note
BeautifulSoup can use parsers such as `html.parser`, `lxml`, or `html5lib`. The notebook defaults to
Python's built-in `html.parser` for portability.
