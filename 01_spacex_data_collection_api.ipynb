# 01 — SpaceX Data Collection API
**IBM Applied Data Science Capstone — Falcon 9 First-Stage Landing Prediction**

This notebook documents the API collection stage. It uses an HTTP GET request, JSON normalization and
filters the launch history to Falcon 9 records for the capstone analysis.

```python
import requests
import pandas as pd

api_url = "https://api.spacexdata.com/v4/launches"
response = requests.get(api_url, timeout=30)
response.raise_for_status()
launches = response.json()
df = pd.json_normalize(launches)

print("HTTP status:", response.status_code)
print("Raw launch records:", len(df))
print(df.head(3))
```
